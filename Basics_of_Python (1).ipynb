# 2. What is an interpreter in Python?
- An interpreter in Python is a program that reads and executes Python code line by line, rather than compiling the entire code at once.

How It Works:

 * The Python interpreter reads the code.

* It translates it into an intermediate form (bytecode).

* The bytecode is then executed by the Python Virtual Machine (PVM).

Advantages of Using an Interpreter:

* Easier Debugging: Since code is executed line by line, errors are caught immediately.
* Platform Independence: Python code runs on multiple platforms without modification.
* No Need for Compilation: Unlike languages like C or Java, Python doesn’t require a separate compilation step.

Popular Python Interpreters:
CPython – Default and most widely used Python interpreter.

PyPy – A faster interpreter with JIT (Just-In-Time) compilation.

Jython – Runs Python on the Java Virtual Machine (JVM).

IronPython – Integrates Python with .NET and C#.

# 5. What is Mutability in Python?

 Mutability refers to whether an object’s state (its data) can be modified after it is created.

* Mutable Objects: Can be changed after creation (e.g., lists, dictionaries, sets).

* Immutable Objects: Cannot be changed after creation (e.g., tuples, strings, integers).

# 6.  Why Are Lists Mutable, but Tuples Are Immutable?

 * Lists (list) are mutable because they are designed for dynamic data storage. You can add, remove, or modify elements.

 * Tuples (tuple) are immutable to ensure data integrity and faster performance. They are useful for fixed collections of items.

# 7.  Difference Between == and is Operators in Python?

* == (Equality Operator): Checks if two values are equal.

* is (Identity Operator): Checks if two variables refer to the same memory location.



# 8. What Are Logical Operators in Python?

# 5. Create a list and tuple in Python, and demonstrate how attempting to change an element works differently for each

ans

```python
# Creating a list (mutable)
my_list = [1, 2, 3]
print("Original List:", my_list)
my_list[0] = 100  # Modifying list
print("Modified List:", my_list)

# Creating a tuple (immutable)
my_tuple = (1, 2, 3)
print("\nOriginal Tuple:", my_tuple)

# Attempting to modify tuple (will cause an error)
try:
    my_tuple[0] = 100
except TypeError as e:
    print("Error:", e)
```

# 7. Write a function to demonstrate the behavior of mutable and immutable arguments

Ans

```python
def demo(a, b):
    a += 10  # Immutable (int) - creates a new copy
    b.append(4)  # Mutable (list) - modifies original list

x = 5  # Immutable
y = [1, 2, 3]  # Mutable

demo(x, y)

print(x)  # Output: 5 (unchanged)
print(y)  # Output: [1, 2, 3, 4] (modified)
```

#14 Write a program to reverse a string using a while loop.

ans

```python
# Take input from user
string = input("Enter a string: ")

# Initialize variables
reversed_string = ""
i = len(string) - 1  # Start from last index

# Loop to reverse the string
while i >= 0:
    reversed_string += string[i]  # Add character to new string
    i -= 1  # Move to previous character

# Print reversed string
print("Reversed String:", reversed_string)
```

# 15. Write a Python program to calculate the factorial of a number provided by the user using a while loop.

ans

```python
# Take input from user
num = int(input("Enter a number: "))

# Initialize variables
factorial = 1
i = num

# Calculate factorial using while loop
while i > 0:
    factorial *= i  # Multiply current number
    i -= 1  # Decrease by 1

# Print result
print("Factorial of", num, "is:", factorial)
```
